```python
from benchmark_utils import benchmark_with_percentiles
import numpy as np
import python_kernel
python_kernel.init()
```

```python
N = 1024 * 1024 * 100
aHost = np.arange(N, dtype=np.float32)
bHost = np.arange(N, dtype=np.float32)

aDevice = python_kernel.create_device_buffer(aHost)
bDevice = python_kernel.create_device_buffer(bHost)
rDevice = python_kernel.create_device_buffer(aHost.size * aHost.itemsize)
```

```python
benchmark_with_percentiles(lambda: python_kernel.add(aDevice, bDevice, rDevice))
```

```python
python_kernel.destroy_device_buffer(aDevice)
python_kernel.destroy_device_buffer(bDevice)
python_kernel.destroy_device_buffer(rDevice)
python_kernel.clear()
```
